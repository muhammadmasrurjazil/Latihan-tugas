```python
import numpy as np

M = np.array([
[2, 1, 1],
[1, 2, 1],
[1, 1, 2],
[2, 1, 3]
])

B = np.array([
[1, 0, 1, 1],
[0, 1, 1, 1],
[1, 1, 0, 1]
])

D = np.array([
[10],
[8],
[6]
])

R = M @ D
Y = B @ R


print("komponen (R) : ")
print(R)

print("\nArea Gudang (Y) : ")
print(Y)
```
